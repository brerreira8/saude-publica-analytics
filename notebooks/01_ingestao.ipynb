```python
# ARQUITETURA: Camada de ingestão (1/5)
# Responsável por carregar os dados brutos para a memória
# Dentro do Google Colab ficam em /content/

# Pandas: bibiloteca principal para manipulação de dados
import pandas as pd # chama o Pandas com o pseudo pd

# os: para montar caminhos de forma segura
import os

# No colab os arquivos upados ficam sempre em /content/
FILE_PATH = "/content/INFLUD24.csv"

print(f"✅ Arquivo esperado em: {FILE_PATH}")
print(f"📁 Arquivo existe: {os.path.exists(FILE_PATH)}")


```

```python
# INGESTÃO: Leitura do CSV com configurações para DATASUS
# Arquivos do governo brasileiro têm padrões específicos

df = pd.read_csv(
    FILE_PATH,
    sep=";",            # DATASUS usa ponto e vírgula como separador
    encoding="latin-1", # Encoding padrão dos arquivos do governo
    low_memory=False    # Evita inferência incorreta de tipos
)

print(f"✅ Dados carregados com sucesso!")
print(f"📊 Linhas: {df.shape[0]:,}")
print(f"📋 Colunas: {df.shape[1]}")
```

```python
# EXPLORAÇÃO INICIAL: Entender o dataset antes de limpar
# Essa análise guia todas as decisões da próxima etapa

# Primeiras linhas do dataset
print("=== Primeiras linhas ===")
print(df.head())

# Tipo de dado de cada coluna
print("\n=== Tipos de dados ===")
print(df.dtypes)

# Top 20 colunas com mais valores nulos
# Essencial para planejar a limpeza
print("\n=== Valores nulos por coluna (top 20) ===")
print(df.isnull().sum().sort_values(ascending=False).head(20))
```
